**Posibilidades hay sin tener en cuenta las restricciones**<br>

Para resolver esta cuestión, primero se analizan los elementos de los que se dispone según el enunciado del problema:
1. Hay 10 partidos de fútbol que se deben programar en una jornada.
2. Existen 10 franjas horarias disponibles a lo largo del fin de semana (1 el viernes, 4 el sábado, 4 el domingo y 1 el lunes).

Dado que el enunciado permite que haya solapamientos (es decir, varios partidos pueden coincidir en la misma franja horaria sin invalidar la solución, aunque se aplique una penalización), se trata de un problema matemático de variaciones con repetición.

Para cada uno de los 10 partidos, se puede elegir cualquiera de los 10 horarios disponibles de forma independiente. Por tanto, el cálculo se formula de la siguiente manera:

$$Variaciones\ con\ repetición = m^n = 10^{10}$$

* Total de posibilidades sin restricciones: 10.000.000.000 (10 mil millones de combinaciones).

**Posibilidades teniendo en cuenta todas las restricciones**

El enunciado indica que existen únicamente dos restricciones obligatorias que invalidarían una solución si no se cumplen:
1. Se debe asignar, al menos, un partido al viernes a las 20h.
2. Se debe asignar, al menos, un partido al lunes a las 20h.

Para calcular matemáticamente las combinaciones válidas, es mucho más sencillo y eficiente utilizar el principio de inclusión-exclusión (calcular los casos totales y restar los casos que rompen las restricciones).

Se definen los siguientes escenarios inválidos:
* Conjunto A (sin viernes): combinaciones donde no se juega ningún partido el viernes. Quedan 9 horarios disponibles para los 10 partidos, por lo que hay $9^{10}$ combinaciones inválidas.
* Conjunto B (sin lunes): combinaciones donde no se juega ningún partido el lunes. Quedan 9 horarios disponibles para los 10 partidos, por lo que hay $9^{10}$ combinaciones inválidas.
* Intersección A y B (ni viernes ni lunes): al restar los conjuntos A y B, se está restando dos veces el escenario donde no hay partidos ni el viernes ni el lunes. Se debe sumar esta intersección una vez para equilibrar matemáticamente la ecuación. Si se quitan ambos días, quedan 8 horarios para 10 partidos, resultando en $8^{10}$ combinaciones.

La fórmula de combinaciones válidas se expresa así:

$$Total\ válidas = Total\ posible - A - B + (A \cap B)$$
$$Total\ válidas = 10^{10} - 9^{10} - 9^{10} + 8^{10}$$

Se realiza el cálculo numérico:
* $10^{10} = 10.000.000.000$
* $9^{10} = 3.486.784.401$
* $8^{10} = 1.073.741.824$

$$Total = 10.000.000.000 - 3.486.784.401 - 3.486.784.401 + 1.073.741.824$$

Total de posibilidades teniendo en cuenta las restricciones: 4.100.172.922 combinaciones válidas.

Para abordar este problema computacionalmente y evaluar las soluciones de forma rápida, la estructura de datos idónea es una combinación de diccionarios y listas, debido a su eficiencia (búsquedas en tiempo constante) y claridad. Se proponen los siguientes elementos.

1. Audiencias base: un diccionario cuyas claves sean tuplas con las categorías de los equipos (ordenadas alfabéticamente) y sus valores sean la audiencia base en millones. Por ejemplo: `audiencias_base = {('A', 'A'): 2.0, ('A', 'B'): 1.3, ('C', 'C'): 0.47, ...}`.
2. Ponderación de horarios: un diccionario donde la clave sea el identificador del horario y el valor sea el coeficiente reductor de audiencia. Por ejemplo: `modificadores_horario = {'Viernes_20': 0.4, 'Sabado_20': 1.0, ...}`.
3. Penalizaciones por coincidencia: una lista simple donde el índice represente el número de coincidencias y el valor sea el porcentaje final de audiencia tras aplicar la reducción. Por ejemplo: `penalizaciones = [1.0, 0.75, 0.55, 0.40, ...]`.

Esta estructura resulta muy útil porque separa por completo los datos estáticos de la variable dinámica. Esto permite programar la función de evaluación con un código limpio y evitar cálculos redundantes al acceder a los datos mediante una simple consulta directa.

**Según el modelo para el espacio de soluciones (*) ¿cuál es la función objetivo?**

La función objetivo es la expresión matemática que calcula la audiencia total de una jornada en base a los horarios asignados al conjunto de partidos, aplicando los modificadores pertinentes. Se formula de la siguiente manera:

$$Audiencia\ Total = \sum_{i=1}^{10} (Base_i \times Ponderacion\_Horario_i \times Penalizacion\_Coincidencia_i)$$

En esta fórmula, las variables que intervienen se definen del siguiente modo:
* $Base_i$: es la audiencia base del partido $i$ según las categorías de los equipos que se enfrentan.
* $Ponderacion\_Horario_i$: es el coeficiente de reducción de audiencia que se aplica por jugar en un día y hora específicos.
* $Penalizacion\_Coincidencia_i$: es el porcentaje de audiencia restante tras aplicar la penalización correspondiente si existen otros partidos programados exactamente a la misma hora.

```
def evaluar_audiencia(solucion, partidos, audiencias_base, ponderaciones, penalizaciones):
    audiencia_total = 0

    # se recorren los 10 partidos programados en el calendario propuesto
    for i in range(len(solucion)):
        horario_asignado = solucion[i]

        # 1. obtener audiencia base según las categorías de los equipos
        base = audiencias_base[partidos[i]['categorias']]

        # 2. obtener modificador multiplicador por el día y hora jugados
        modificador = ponderaciones[horario_asignado]

        # 3. calcular penalización contando cuántos partidos coinciden a esa misma hora
        # (se resta 1 para que el partido no se cuente a sí mismo como coincidencia)
        coincidencias = solucion.count(horario_asignado) - 1
        penalizacion = penalizaciones[coincidencias]

        # se aplica la fórmula matemática de la función objetivo
        audiencia_partido = base * modificador * penalizacion

        # se acumula al total de la jornada
        audiencia_total += audiencia_partido

    return audiencia_total

def fuerza_bruta_liga(partidos, horarios, audiencias_base, ponderaciones, penalizaciones):
    mejor_audiencia = 0
    mejor_solucion = None

    # generación exhaustiva mediante bucles anidados para los 10 partidos
    for p1 in horarios:
        for p2 in horarios:
            for p3 in horarios:
                for p4 in horarios:
                    for p5 in horarios:
                        for p6 in horarios:
                            for p7 in horarios:
                                for p8 in horarios:
                                    for p9 in horarios:
                                        for p10 in horarios:

                                            # se agrupan los horarios asignados en una lista
                                            solucion = [p1, p2, p3, p4, p5, p6, p7, p8, p9, p10]

                                            # comprobar restricciones obligatorias
                                            if 'Viernes_20' not in solucion or 'Lunes_20' not in solucion:
                                                continue

                                            # evaluar la función objetivo
                                            audiencia_actual = evaluar_audiencia(solucion, partidos, audiencias_base, ponderaciones, penalizaciones)

                                            # actualizar si es la mejor hasta ahora
                                            if audiencia_actual > mejor_audiencia:
                                                mejor_audiencia = audiencia_actual
                                                mejor_solucion = solucion

    return mejor_solucion, mejor_audiencia

# Las variables se definirían:

# horarios = [
#     'Viernes_20',
#     'Sabado_12', 'Sabado_16', 'Sabado_18', 'Sabado_20',
#     'Domingo_12', 'Domingo_16', 'Domingo_18', 'Domingo_20',
#     'Lunes_20'
# ]

# partidos = [
#     {'partido': 'Celta - Real Madrid', 'categorias': 'B_A'},
#     {'partido': 'Valencia - R. Sociedad', 'categorias': 'B_A'},
#     {'partido': 'Mallorca - Eibar', 'categorias': 'C_C'},
#     {'partido': 'Athletic - Barcelona', 'categorias': 'B_A'},
#     {'partido': 'Leganés - Osasuna', 'categorias': 'C_C'},
#     {'partido': 'Villarreal - Granada', 'categorias': 'B_C'},
#     {'partido': 'Alavés - Levante', 'categorias': 'B_B'},
#     {'partido': 'Espanyol - Sevilla', 'categorias': 'B_B'},
#     {'partido': 'Betis - Valladolid', 'categorias': 'B_C'},
#     {'partido': 'Atlético - Getafe', 'categorias': 'B_B'}
# ]

# audiencias_base = {
#     'A_A': 2.0,
#     'A_B': 1.3, 'B_A': 1.3,
#     'A_C': 1.0, 'C_A': 1.0,
#     'B_B': 0.9,
#     'B_C': 0.75, 'C_B': 0.75,
#     'C_C': 0.47
# }

# ponderaciones = {
#     'Viernes_20': 0.4,
#     'Sabado_12': 0.55,
#     'Sabado_16': 0.7,
#     'Sabado_18': 0.8,
#     'Sabado_20': 1.0,
#     'Domingo_12': 0.45,
#     'Domingo_16': 0.75,
#     'Domingo_18': 0.85,
#     'Domingo_20': 1.0,
#     'Lunes_20': 0.4
# }

# penalizaciones = {
#     0: 1.0,   # 0% de reducción
#     1: 0.75,  # 25% de reducción
#     2: 0.55,  # 45% de reducción
#     3: 0.40,  # 60% de reducción
#     4: 0.30,  # 70% de reducción
#     5: 0.25,  # 75% de reducción
#     6: 0.22,  # 78% de reducción
#     7: 0.20,  # 80% de reducción
#     8: 0.20,  # 80% de reducción
#     9: 0.20   # 80% de reducción
# }
```

Respuesta

Al anidar 10 bucles iterando sobre 10 horarios posibles, el número total de combinaciones generadas se multiplica en cada paso, resultando en un coste de $m^{n}$, es decir, $10^{10}$ combinaciones.

Además, por cada solución que resulta ser válida, el algoritmo se ve obligado a recorrer la lista de 10 partidos para calcular la suma de la audiencia y aplicar las penalizaciones, lo que añade un coste de tiempo lineal *O(n)* por cada combinación.

La complejidad teórica total del algoritmo se define así: *O(n⋅$m^{n}$)*.

Aplicado a los datos concretos de este problema, la complejidad dictamina que se tendrían que realizar decenas de miles de millones de operaciones básicas.
